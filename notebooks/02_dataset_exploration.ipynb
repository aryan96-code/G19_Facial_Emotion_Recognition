## 0. Setup — Paths and Imports

```python
import sys, os
sys.path.insert(0, '/content/G19_Facial_Emotion_Recognition')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_fer2013, EMOTION_LABELS, NUM_CLASSES
from src.utils import set_random_seeds, plot_class_distribution, display_sample_images

set_random_seeds(42)

DATASET_CSV = '/content/G19_Facial_Emotion_Recognition/dataset/fer2013.csv'
RESULTS_DIR = '/content/G19_Facial_Emotion_Recognition/results'
os.makedirs(RESULTS_DIR, exist_ok=True)
print('Setup complete.')
```

## 1. Load Dataset

```python
X_train, y_train, X_val, y_val, X_test, y_test = load_fer2013(DATASET_CSV)

print('\n--- Dataset Shapes ---')
print(f'Train   : X={X_train.shape}  y={y_train.shape}')
print(f'Val     : X={X_val.shape}    y={y_val.shape}')
print(f'Test    : X={X_test.shape}   y={y_test.shape}')
print(f'\nPixel dtype  : {X_train.dtype}')
print(f'Value range  : [{X_train.min():.3f}, {X_train.max():.3f}]')
```

## 2. Class Distribution — Training Set

```python
plot_class_distribution(
    y_train,
    split_name='Training',
    save_path=f'{RESULTS_DIR}/class_distribution_train.png'
)
plt.savefig(f'{RESULTS_DIR}/class_distribution_train.png', dpi=120)
plt.show()

# Print counts
print('\nClass counts (Training):')
for i in range(NUM_CLASSES):
    count = int((y_train == i).sum())
    print(f'  {EMOTION_LABELS[i]:<10} : {count:>5} samples ({count/len(y_train)*100:.1f}%)')
```

## 3. Sample Images per Emotion Class

```python
# Display 5 samples per class
display_sample_images(
    X_train, y_train,
    n_per_class=5,
    save_path=f'{RESULTS_DIR}/sample_images.png'
)

# Show inline
from PIL import Image
img = Image.open(f'{RESULTS_DIR}/sample_images.png')
plt.figure(figsize=(12, 10))
plt.imshow(img)
plt.axis('off')
plt.title('Sample Images per Emotion Class', fontsize=14)
plt.show()
```

## 4. Pixel Intensity Distributions

```python
# Compare pixel intensity histograms across classes
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
axes = axes.flatten()

for class_idx in range(NUM_CLASSES):
    class_imgs = X_train[y_train == class_idx].flatten()
    axes[class_idx].hist(class_imgs, bins=50, color='steelblue', alpha=0.75)
    axes[class_idx].set_title(EMOTION_LABELS[class_idx], fontsize=11)
    axes[class_idx].set_xlabel('Pixel intensity [0,1]')
    axes[class_idx].set_ylabel('Frequency')
    axes[class_idx].grid(True, alpha=0.3)

# Hide unused subplot
axes[-1].set_visible(False)

plt.suptitle('Pixel Intensity Distributions by Emotion', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/pixel_distributions.png', dpi=120)
plt.show()
print(f'Saved → {RESULTS_DIR}/pixel_distributions.png')
```

## 5. Mean Face per Emotion

```python
# Compute and display the mean face image for each emotion
fig, axes = plt.subplots(1, NUM_CLASSES, figsize=(14, 3))

for class_idx in range(NUM_CLASSES):
    class_imgs = X_train[y_train == class_idx, :, :, 0]  # (N, 48, 48)
    mean_face  = class_imgs.mean(axis=0)
    axes[class_idx].imshow(mean_face, cmap='gray', vmin=0, vmax=1)
    axes[class_idx].set_title(EMOTION_LABELS[class_idx], fontsize=10)
    axes[class_idx].axis('off')

plt.suptitle('Mean Face Image per Emotion', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/mean_faces.png', dpi=120)
plt.show()
print(f'Saved → {RESULTS_DIR}/mean_faces.png')
```
